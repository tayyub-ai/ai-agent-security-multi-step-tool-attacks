```python
from huggingface_hub import hf_hub_download

default_repo="unsloth/gemma-4-26B-A4B-it-GGUF"
default_file="gemma-4-26B-A4B-it-UD-Q4_K_M.gguf"

model_path = hf_hub_download(repo_id=default_repo, filename=default_file)
```

```python
import kagglehub

# Replace with path to directory containing model files.
LOCAL_MODEL_DIR = model_path

MODEL_SLUG = default_repo.replace('/', '--')
VARIATION_SLUG = default_file.strip('gemma-4-').strip('.gguf')

kagglehub.model_upload(
  handle = f"mccocoful/{MODEL_SLUG}/gguf/{VARIATION_SLUG}",
  local_model_dir = LOCAL_MODEL_DIR,
  version_notes = 'Update 2026-08-17')
```
